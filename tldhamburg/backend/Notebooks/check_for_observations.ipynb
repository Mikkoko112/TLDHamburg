```python
from datetime import datetime,timedelta
from tldhamburg.backend.Things import Things,DBMaintainer,DataFetcher
import sqlite3
import requests
import time
import json
import os
from concurrent.futures import ThreadPoolExecutor
```

```python

with open("observations_all_datastreams.json","r") as file:
    data = json.load(file)
```

```python
empty_ids = list()
for key,value in data.items():
    if not value:
        empty_ids.append(key)
```

```python
path = f"{os.getcwd()}"
name = f"empty_ids.txt"
file_path = os.path.join(path,name)

with open(file_path,"w") as file:
    for id in empty_ids:
        file.write(str(id) + "\n")
```
