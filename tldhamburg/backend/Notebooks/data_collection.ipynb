```python
import folium, json, pickle, copy
from IPython.display import display
import requests
from tldhamburg.backend.DataFetcher import DataFetcher
data_fetcher = DataFetcher()
```

```python

endpoint = "Things"
params = {"$expand": "Locations"}
data_fetcher.fetch_data(endpoint, params)
```

```python
things_locations = dict()

things_no_locations = list()

for things_data in data_fetcher.data:
    thing_id = things_data["@iot.id"]
    locations_data = things_data["Locations"]
    # print(locations_data)

    if len(locations_data) > 0:
        locations_data = locations_data[0]
        # In case of key error
        if "geometry" not in locations_data["location"].keys():
            print(f"No location data available for Thing ID: {thing_id}")
            things_no_locations.append(thing_id)
            continue

        coordinates = locations_data["location"]["geometry"]["coordinates"]
        # If found, collect the coordinates
        if coordinates:
            things_locations[thing_id] = coordinates
    else:
        print(f"No location data available for Thing ID: {thing_id}")
        things_no_locations.append(thing_id)


len(things_no_locations)
```

```python
thing_locations_copy = copy.deepcopy(things_locations)
thing_nolocations_copy = copy.deepcopy(things_no_locations)

with open("things_locationsx.pkl", "wb") as file:
    pickle.dump(thing_locations_copy, file)

with open("things_no_locationsx.pkl", "wb") as file:
    pickle.dump(thing_nolocations_copy, file)

print("Saved successfully!")
```
